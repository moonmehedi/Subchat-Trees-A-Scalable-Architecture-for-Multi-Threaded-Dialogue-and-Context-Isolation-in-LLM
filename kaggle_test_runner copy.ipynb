```python
# Cell 1: Install vLLM dependencies + ROUGE/BLEU for recall probes
print("="*60)
print("📦 INSTALLING vLLM DEPENDENCIES + RECALL PROBE LIBS")
print("="*60)

# Kaggle often imports NumPy before notebook cells run. Do not downgrade or
# reinstall NumPy in a live kernel; that creates the dtype-size ABI error.
# Also do not reinstall Pillow here. This text-only runner removes torchvision
# so transformers will not walk into fragile torchvision/PIL paths.
import importlib.metadata
from pathlib import Path
import shutil
import subprocess
import sys
import os

os.environ["TRANSFORMERS_NO_TORCHVISION"] = "1"
os.environ["TRANSFORMERS_NO_VISION"] = "1"

try:
    installed_numpy = importlib.metadata.version("numpy")
except importlib.metadata.PackageNotFoundError:
    installed_numpy = None

constraint_args = []
if installed_numpy:
    constraints_path = Path("/kaggle/working/subchat_numpy_constraints.txt")
    constraints_path.write_text(f"numpy=={installed_numpy}\n")
    constraint_args = ["-c", str(constraints_path)]
    print(f"🔒 Preserving existing NumPy {installed_numpy}")


def pip_install(*packages, check=True):
    if shutil.which("uv"):
        cmd = ["uv", "pip", "install", "-q", "--system", "--upgrade"]
    else:
        cmd = [sys.executable, "-m", "pip", "install", "-q", "--upgrade"]
    cmd.extend(packages)
    cmd.extend(constraint_args)
    print("$", " ".join(cmd))
    return subprocess.run(cmd, check=check)


def pip_uninstall(*packages):
    if shutil.which("uv"):
        cmd = ["uv", "pip", "uninstall", "-q", "--system", *packages]
    else:
        cmd = [sys.executable, "-m", "pip", "uninstall", "-y", "-q", *packages]
    print("$", " ".join(cmd))
    return subprocess.run(cmd, check=False)


pip_uninstall("tensorflow", "torchvision")
pip_install("vllm", "triton==3.2.0", "logits-processor-zoo")
pip_install("rouge-score>=0.1.2", "nltk>=3.8.0")

# Verify NumPy's compiled extension is internally consistent.
import numpy as np
import numpy.random as _np_random
_np_random.RandomState(0)

# Download NLTK tokenizer data (needed for BLEU)
import nltk
nltk.download('punkt_tab', quiet=True)

print(f"✅ Dependencies installed; NumPy ABI OK ({np.__version__})")
print("="*60)
```

```python
# Cell 2: Import libraries
import importlib.abc
import importlib.util
import os
import shutil
import subprocess
import sys

# This is a text-only model runner. torchvision/PIL is not needed, and this
# Kaggle account has a broken PIL stack. Block vision imports before vLLM loads.
os.environ["TRANSFORMERS_NO_TORCHVISION"] = "1"
os.environ["TRANSFORMERS_NO_VISION"] = "1"

# If a previous interactive attempt partially imported these modules, remove
# them so this cell can recover without a manual kernel restart.
for name in list(sys.modules):
    if (
        name == "vllm" or name.startswith("vllm.") or
        name == "transformers" or name.startswith("transformers.") or
        name == "torchvision" or name.startswith("torchvision.") or
        name == "PIL" or name.startswith("PIL.")
    ):
        del sys.modules[name]

class _BlockVisionImports(importlib.abc.MetaPathFinder):
    def find_spec(self, fullname, path=None, target=None):
        if fullname == "torchvision" or fullname.startswith("torchvision."):
            raise ModuleNotFoundError("torchvision disabled for text-only vLLM run")
        return None

if not any(isinstance(finder, _BlockVisionImports) for finder in sys.meta_path):
    sys.meta_path.insert(0, _BlockVisionImports())

original_find_spec = importlib.util.find_spec

def _no_vision_find_spec(name, package=None):
    if name == "torchvision" or name.startswith("torchvision"):
        return None
    return original_find_spec(name, package)

importlib.util.find_spec = _no_vision_find_spec

try:
    from transformers.utils import import_utils as _tf_import_utils
    _tf_import_utils._torchvision_available = False
    _tf_import_utils._vision_available = False
except Exception as exc:
    print(f"⚠️  Could not pre-disable transformers vision imports: {exc}")

import numpy as np

# Fail early with a clear message if NumPy was already broken by a prior run.
try:
    import numpy.random as _np_random
    _np_random.RandomState(0)
except ValueError as exc:
    raise RuntimeError(
        "NumPy binary ABI mismatch detected. In Kaggle Save & Run, rerun the "
        "notebook from a fresh run after Cell 1 has been updated."
    ) from exc

import torch
import vllm
from logits_processor_zoo.vllm import MultipleChoiceLogitsProcessor

print("="*60)
print("📚 LIBRARIES IMPORTED")
print("="*60)
print(f"✅ NumPy version: {np.__version__}")
print(f"✅ PyTorch version: {torch.__version__}")
print(f"✅ CUDA available: {torch.cuda.is_available()}")
print(f"🎮 GPUs available: {torch.cuda.device_count()}")
if torch.cuda.is_available():
    for i in range(torch.cuda.device_count()):
        print(f"   GPU {i}: {torch.cuda.get_device_name(i)}")
print("="*60)
```

```python
# Cell 4: Load vLLM model on Kaggle GPUs (takes 2-3 minutes)
print("="*60)
print("🚀 LOADING vLLM MODEL")
print("="*60)
print(f"📂 Model: {model_path}")
print(f"🎮 GPUs: {torch.cuda.device_count()}")
print("⏳ This takes 2-3 minutes...")
print("="*60)

# Keep transformers vision disabled before touching tokenizer utilities.
os.environ["TRANSFORMERS_NO_TORCHVISION"] = "1"
os.environ["TRANSFORMERS_NO_VISION"] = "1"
try:
    from transformers.utils import import_utils as _tf_import_utils
    _tf_import_utils._torchvision_available = False
    _tf_import_utils._vision_available = False
except Exception as exc:
    print(f"⚠️  Could not keep transformers vision disabled: {exc}")

# Compatibility fix for vLLM 0.8.x + some Transformers/Qwen tokenizer combos.
# This vLLM version only supports tokenizer_mode: auto, slow, mistral, custom.
# It may still ask slow Qwen tokenizers for all_special_tokens_extended, so add
# that property before vLLM initializes its tokenizer cache.
from transformers.tokenization_utils_base import PreTrainedTokenizerBase


def _all_special_tokens_extended(self):
    return getattr(self, "all_special_tokens", [])

PreTrainedTokenizerBase.all_special_tokens_extended = property(
    _all_special_tokens_extended
)
print("✅ Tokenizer compatibility shim installed")

llm = vllm.LLM(
    model_path,
    tokenizer=model_path,
    tokenizer_mode="auto",
    quantization='awq',
    tensor_parallel_size=torch.cuda.device_count(),
    gpu_memory_utilization=0.91,
    trust_remote_code=True,
    dtype="half",
    enforce_eager=True,
    max_model_len=5000,
    disable_log_stats=True,
    enable_prefix_caching=True
)
tokenizer = llm.get_tokenizer()

print("\n" + "="*60)
print("✅ vLLM MODEL LOADED SUCCESSFULLY!")
print("="*60)
print(f"   Memory per GPU: ~{torch.cuda.get_device_properties(0).total_memory / 1024**3 * 0.91:.1f}GB used")
print("="*60)
```

```python
# Cell 9: RUN BUFFER TESTS (SERVERLESS - no HTTP server needed!)
# Running on ENTIRE DATASET with robust error handling

print("="*60)
print("🚀 RUNNING SERVERLESS BUFFER TESTS - FULL DATASET")
print("="*60)
print("📊 Testing buffer sizes: 5, 10, 15, 20")
print("📄 Dataset: merged_realistic_interleaved.json (VBRI interleaved, 309 turns)")
print("📤 Results will auto-push to GitHub after each buffer")
print("⏳ This may take several hours")
print("✅ NO SERVER NEEDED - using direct Python imports!")
print("✅ Max output tokens: 300")
print("="*60)

os.chdir("/kaggle/working/Subchat-Trees/backend")

# Import and run the serverless test runner
import sys
import glob
from pathlib import Path
sys.path.insert(0, "/kaggle/working/Subchat-Trees/backend")
sys.path.insert(0, "/kaggle/working/Subchat-Trees/backend/dataset")

from dataset.kaggle_serverless_runner import ServerlessTestRunner

# Create runner
runner = ServerlessTestRunner()
runner.test_mode = "both"  # Run both baseline and system tests

# RUN ONLY THE MERGED INTERLEAVED DATASET
scenario_dir = Path("/kaggle/working/Subchat-Trees/backend/dataset/scenarios")
scenario_files = ["merged_realistic_interleaved.json"]

# Buffer sizes to test
buffer_sizes = [20]

print(f"\n📄 Discovered {len(scenario_files)} scenario files:")
for idx, file in enumerate(scenario_files, 1):
    print(f"   {idx}. {file}")
print(f"\n📦 Buffer sizes: {buffer_sizes}")
print(f"\n⏱️  Estimated time: ~{len(scenario_files) * len(buffer_sizes) * 15} minutes")
print("="*60)

# Validate scenarios before starting (quick check)
print("\n🔍 Validating scenario files...")
import json
valid_scenarios = []
for scenario_file in scenario_files:
    try:
        scenario_path = scenario_dir / scenario_file
        with open(scenario_path, 'r') as f:
            data = json.load(f)
            # Check for BOTH formats: "conversation" (old) and "conversations" (structured)
            if ("conversation" in data and isinstance(data["conversation"], list)) or \
               ("conversations" in data and isinstance(data["conversations"], list)):
                conv_key = "conversation" if "conversation" in data else "conversations"
                valid_scenarios.append(scenario_file)
                print(f"   ✅ {scenario_file} ({len(data[conv_key])} steps)")
            else:
                print(f"   ⚠️  {scenario_file} - skipping (invalid format)")
    except Exception as e:
        print(f"   ❌ {scenario_file} - skipping (error: {e})")

print(f"\n✅ {len(valid_scenarios)}/{len(scenario_files)} scenarios validated")
print("="*60)

if not valid_scenarios:
    print("❌ ERROR: No valid scenario files found!")
else:
    # Run the comparison with validated scenarios
    try:
        runner.run_buffer_comparison(
            valid_scenarios,
            buffer_sizes=buffer_sizes
        )
        print("\n" + "="*60)
        print("🎉 ALL TESTS COMPLETED SUCCESSFULLY!")
        print("="*60)
    except KeyboardInterrupt:
        print("\n⚠️  Tests interrupted by user")
    except Exception as e:
        print(f"\n❌ ERROR during test execution: {e}")
        import traceback
        traceback.print_exc()
```
