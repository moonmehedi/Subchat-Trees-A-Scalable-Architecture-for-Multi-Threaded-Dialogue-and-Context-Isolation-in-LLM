```python
# write a infinte loop that counts from 1 to 10000 and again from 10000 to 1 and prints a number after every 500 counds

while True:
    # Count up from 1 to 10000
    for i in range(1, 10001):
        if i % 500 == 0:
            print(i)

    # Count down from 10000 to 1
    for i in range(10000, 0, -1):
        if i % 500 == 0:
            print(i)
```
